# Def + Prompt

```python
grammar = r"""
start: number operator number

number: /-?[0-9]+(\.[0-9]*)?/
operator: "+" | "-" | "*" | "/"

%ignore /[ \t]+/
"""

print(grammar)
```

```python
def build_prompt(textual_query):
    return f"""
Extract the two numeric operands and the arithmetic operator
from the user's question.

Return ONLY an expression in this format:
operand operator operand

Use these operators:
+ for addition
- for subtraction
* for multiplication
/ for division

Do not calculate the answer. Preserve the operand order.

Examples:

Question: What is 45.1 plus 23.54?
Extraction: 45.1 + 23.54

Question: What is 327. multiplied by 11.0?
Extraction: 327. * 11.0

Question: What is 9.25 minus 3?
Extraction: 9.25 - 3

Question: What is 120.4 divided by 4.0?
Extraction: 120.4 / 4.0

Now extract from this question:
Question: {textual_query}
Extraction:
"""
```

```python
# initiliaze syncode
model_name = "microsoft/phi-2"

syn_llm = Syncode(
    model=model_name,
    grammar=grammar,
    parse_output_only=True,
    max_new_tokens=30
)

print("SynCode model initialized!")
```

```python
test_query = "What is 327. multiplied by 11.0?"

prompt = build_prompt(test_query)
output = syn_llm.infer(prompt)[0]

print("Question:", test_query)
print("Extracted expression:", repr(output))
```

```python
import re

def run_syncode(textual_query):
    """
    Extract two operands and an operator using SynCode,
    calculate the result using Python, and return:
    (ordered operands, operator, final answer)
    """
    prompt = build_prompt(textual_query)
    output = syn_llm.infer(prompt)[0]

    # Extract the two operands and the operator.
    pattern = r"(-?\d+(?:\.\d*)?)\s*([+\-*/])\s*(-?\d+(?:\.\d*)?)"
    match = re.fullmatch(pattern, output.strip())

    if match is None:
        raise ValueError(f"Could not parse SynCode output: {output!r}")

    first_operand = float(match.group(1))
    operator = match.group(2)
    second_operand = float(match.group(3))

    # Calculate the result using Python, not the language model.
    if operator == "+":
        answer = first_operand + second_operand
    elif operator == "-":
        answer = first_operand - second_operand
    elif operator == "*":
        answer = first_operand * second_operand
    elif operator == "/":
        if second_operand == 0:
            raise ZeroDivisionError("Cannot divide by zero.")
        answer = first_operand / second_operand
    else:
        raise ValueError(f"Unsupported operator: {operator}")

    return [first_operand, second_operand], operator, float(answer)
```

```python
test_queries = [
    "What is 327. multiplied by 11.0?",
    "What is 45.1 plus 23.54?",
    "What is 120.4 divided by 4.0?",
    "What is 9.25 minus 3?",
]

for query in test_queries:
    operands, operator, answer = run_syncode(query)
    print(f"Question: {query}")
    print(f"Operands: {operands}, Operator: {operator}, Answer: {answer}")
    print()
```

```python
# adding automated tests and allat
import math

tests = [
    ("What is 327. multiplied by 11.0?", [327.0, 11.0], "*", 3597.0),
    ("What is 45.1 plus 23.54?", [45.1, 23.54], "+", 68.64),
    ("What is 120.4 divided by 4.0?", [120.4, 4.0], "/", 30.1),
    ("What is 9.25 minus 3?", [9.25, 3.0], "-", 6.25),
]

for query, expected_operands, expected_operator, expected_answer in tests:
    operands, operator, answer = run_syncode(query)

    assert len(operands) == 2
    assert all(
        math.isclose(actual, expected, rel_tol=1e-9, abs_tol=1e-9)
        for actual, expected in zip(operands, expected_operands)
    )
    assert operator == expected_operator
    assert math.isclose(answer, expected_answer, rel_tol=1e-9, abs_tol=1e-9)

print("All tests pass!!!")
```
